```python
#for loop = count vowel 
a = input (" Enter a string : ")
print(a)
count = 0
for i in a :
    if i .lower() in "aeiou":# it was necessary to convert in lower case if the word had both lower and upper case 
        count+=1
print("Number of vowels : ", count )
    
```

```python
#string method = Palindrome check 
text = input("Enter a string: ")

clean = text.replace(" ", "").lower()

if clean == clean[::-1]:#it checks the clean code .. -1 reverse string same as the clean 
    print("Palindrome")
else:
    print("Not Palindrome")
```
