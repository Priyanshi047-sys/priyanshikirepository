```python
'''recursion and set'''
#recursion = calling the function again and again 
#FACTORIAL 
n = int(input("enter a value : "))
def factorial(n):
    if(n==0 or n==1):
        return 1
    else:
        return n * factorial(n-1) 
print(factorial(n))
```

```python
#fibonacci series 
n = int(input("no . of terms  : "))
def fibonacci(n):
    if(n<=1):
        return n 
    else :
        return fibonacci(n-1) + fibonacci(n-2)
print("fibonacci series : ")
for i in range(n) : 
    print(fibonacci(i), end = " ")
        
```
