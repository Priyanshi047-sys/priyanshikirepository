```python
#dictionary = to store many values in one variable 
#dictonaries are ordered collection of data items 
info = {'name': 'karan','age':19,'eligible':True}
print(info)
print(info.keys())#same no. of  keys 
print(info.values())#same no. of values 
for key in info.keys():
    print(info[key])
```

```python
'''exception handling '''
a = input("enter the number : ")
print(f"Multiplication table of {a} is: ")
try:
    for i in range (1,11):#it is a process of responding to unwanted events when a computer program runs 
        print(f"{int(a)}X {i} = {int(a)*i}")
except:
    print("Invalid Input!")

print("Some imp lines of code ")
print("End of program")

```

```python
try: 
    num = int(input("Enter an integer: "))
    a = [6,3]
    print(a[num])# give index no. and u will get the value = also called list indexing 
except ValueError:
    print("Number entered is not an integer.")

except IndexError:
    print("Index Error")
```

```python
'''finally keyword'''
#can we used with function ..this  code will be execusted at any cost 
def func1():
    try:
        l = [1,5,6,7]
        i = int(input("Enter the index: "))
        print(l[i])
        return 1
    except:
        print("Some error occured")
        return 0 
    finally:
        print("I am always executed")
x = func1()
print(x)
```
