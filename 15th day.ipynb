```python
'''import'''
# importing can be used to import diffferent packages of python module into ur current python project 
import math 
result = math.sqrt(9)
print(result)
```

```python
# from keyword can be used to import specific functions or variables from a module using the from keyword .
from math import sqrt ,pi 
result = sqrt(9)#now there is no need to write import 
print(result)
print(pi)
```

```python
# as keyword 
import math as m 
result = m.sqrt(9)
print(result)
print(m.pi)
```

```python
#dir function can be used to see all the function present inside the module 
import math
print(dir(math))
print(math.nan,type(math.nan))
```
