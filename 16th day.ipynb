```python
# local variable and global variable 
# A local variable is a  variable that is defined within a function
# ans is only accessible within that function.It is created when the function is called and destroyed when the function returns 
# global variable is a variable that is defined outside of a function 
# and is accessible from within any function in your code  
```

```python
x = 10#global variable 

def my_function():
    y = 5#local variable 
    print(y)

my_function()
print(x)
print(y)#this will cause error cause y is a local variable and is not accessible 
#outside of the function
```

```python
x = 10#global variable 

def my_function():
    global x#but still there is no need to convert global variable
    x = 4
    y = 5#local variable 
    print(y)

my_function()
print(x)
```

```python
f = open('myfile.txt','r')
while True:
    line = f.readline()#read the file line by line
    if not line:
        break
    print(line)
```

```python
#seek() function allows you to move the current position within a file to a specific file 
with open('myfile.txt','r') as f:
    print(type(f))
    f.seek(10)
   
    print(f.tell())#tells at which position in file we are
    data = f.read(5)#read the next 5 byte 
    print(data)
```
