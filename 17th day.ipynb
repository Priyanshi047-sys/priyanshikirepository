```python
def cube(x):
    return x*x*x
print(cube(2))
l = [2,3,4,5,6]
newl = list(map(cube,l))
print(newl)
```

```python
#filter function = to filter out from the given list or sequences
#syntax = filter(predicate,iterable)
def filter_function(a):
    return a >3
newnewl = list(filter(filter_function,l))
print(newnewl)
```

```python
#snake water gun game 
import random

# Function to decide winner
def check_winner(player, computer):
    if player == computer:
        return "Draw"

    # Winning conditions for player
    elif (player == "S" and computer == "W") or \
         (player == "W" and computer == "G") or \
         (player == "G" and computer == "S"):
        return "Player Wins"

    else:
        return "Computer Wins"


# Main program
def play_game():
    choices = ["S", "W", "G"]

    computer = random.choice(choices)
    player = input("Enter your choice (S for Snake, W for Water, G for Gun): ").upper()

    if player not in choices:
        print("Invalid input!")
        return

    print("Computer chose:", computer)

    result = check_winner(player, computer)
    print("Result:", result)


# Run the game
play_game()
```
