```python
#typecasting
'''basically conversion one data type into another '''
a="1"
b="1"
c=a+b
print(c)
d=(int(a)+int(b))
print(d)
#here string is converted in to integer 
```

```python
'''explicit typecasting is basically conversion by developer or us manually '''
string = "15"
number = 7
string_number = int(string) #throws an error if string is not a valid integer 
sum = number + string_number
print("the sum of both the numer is : ",sum)
```

```python
'''implicit type casting is what is done automatically = converts data type to higher data type to prevent data loss '''
a = 7
print(type(a))
b= 3.0
print(type(b))
c= a+b
print(c)
print(type(c))
```
