```python
#example 1
for i in range(12):
    if(i==10):
        print("skip the iteration ")
        continue 
    print("5 X", i , "=", 5*i)# continue directly jumps to the next iteration/ comes out of the iteration 

#example 2
for i in [2,3,4,6,8,0]:
    if(i%2!=0):
        continue
    print(i)
```
