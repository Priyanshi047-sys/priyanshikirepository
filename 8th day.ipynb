```python
'''List comprehension 
List = [Expression(item) for item in iterable if Condition]'''
names = ["Milo", "Sarah", "Bruno", "Anastasia", "Rosa"]
namesWith_O = [item for item in names if "o" in item]
print(namesWith_O)

names = ["Milo", "Sarah", "Bruno", "Anastasia", "Rosa"]
namesWith_O = [item for item in names if (len(item) > 4)]
print(namesWith_O)
```

```python
'''list methods'''
l = [11,45,6,7,86]
l.append(7)#append adds the element at the end
print(l)
l.sort()#sort does sorting in the asscending order
print(l)
l.sort(reverse=True)#it does sorting in the descending order
print(l)
l.reverse()#it reverse the list
print(l)
print(l.count(7))
print(l.index(7))#index of 7 = first occurence
#copy,insert,and extend bacha h 
k = l+marks
print(k)
```
