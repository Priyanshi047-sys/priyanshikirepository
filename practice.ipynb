```python
x = int(input("enter a number : "))
# x is the variable to match
match x:
    # if x is 0
    case 0:
        print("x is zero")
    # case with if-condition
    case 4 if x % 2 == 0:
        print("x % 2 == 0 and case is 4")
    # Empty case with if-condition
    case _ if x < 10:
        print("x is < 10")
    
    # default case(will only be matched if the above cases were not matched)
    # so it is basically just an else:
    case _:
        print(x)
        
```

```python
# kbc type question , answer and amount taking home 
questions = [
    ["What is the capital of India?", "A. Delhi", "B. Mumbai", "C. Kolkata", "D. Chennai", "A"],
    ["2 + 2 = ?", "A. 3", "B. 4", "C. 5", "D. 6", "B"],
    ["Which language is used for DSA?", "A. HTML", "B. CSS", "C. C++", "D. Photoshop", "C"]
]

prize_money = [1000, 5000, 10000]

money = 0

for i in range(len(questions)):
    q = questions[i]

    print("\nQuestion for ₹", prize_money[i])
    print(q[0])
    print(q[1])
    print(q[2])
    print(q[3])
    print(q[4])

    answer = input("Enter your answer (A/B/C/D): ").upper()

    if answer == q[5]:
        print("✅ Correct!")
        money += prize_money[i]
    else:
        print("❌ Wrong answer!")
        break

print("\nYou won ₹", money)
```

```python
#fibonacci series 
n = int(input("no . of terms  : "))
def fibonacci(n):
    if(n<=1):
        return n 
    else :
        return fibonacci(n-1) + fibonacci(n-2)
print("fibonacci series : ")
for i in range(n) : 
    print(fibonacci(i), end = " ")
        
```

```python
# secret code 
# # Coding Rules (Encoding):
# If the word contains at least 3 characters:
# Remove the first letter and append it at the end
# Now append three random characters at the starting and the end
# Else:
# Simply reverse the string
# 🔓 Decoding Rules:
# If the word contains less than 3 characters:
# Reverse it
# Else:
# Remove 3 random characters from the start and end
# Now remove the last letter and append it at the beginning
```

```python
def encode(word):
    if len(word) >= 3:
        word = word[1:] +word[0]#move first letter to end 
        return "abc" + word + "xyz"#fixed character instead of random
    else:
        return word[::-1]#reverse
#decode
def decode(word):
    if len(word)<3:
        return word[::-1]
    else:
        word = word[3:-3]#remove abc and xyz
        return word[-1]+word[:-1]#move first letter to front
# Main
choice = input("e for encode, d for decode: ")
text = input("Enter message: ")

words = text.split()
result = []

for w in words:
    if choice == "e":
        result.append(encode(w))
    else:
        result.append(decode(w))

print("Result:", " ".join(result))
```
